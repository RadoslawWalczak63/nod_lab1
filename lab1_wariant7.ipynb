## Krok 40. Łączenie ramek danych – merge i concat
`merge()` łączy ramki po wspólnej kolumnie, `concat()` skleja je pod sobą lub obok siebie.

```python
srednie = df.groupby("location", as_index=False)["val"].mean().rename(columns={"val": "srednia_kraju"})
polaczone = pd.merge(df, srednie, on="location", how="left")
display(polaczone.head())

rok_1990 = df[df["year"] == 1990].head(3)
rok_2019 = df[df["year"] == 2019].head(3)
pd.concat([rok_1990, rok_2019], ignore_index=True)
```

## Krok 41. Nowe kolumny za pomocą operacji matematycznych

```python
polaczone["roznica"] = polaczone["val"] - polaczone["srednia_kraju"]
polaczone["stosunek"] = polaczone["val"] / polaczone["srednia_kraju"]
polaczone[["location", "year", "val", "srednia_kraju", "roznica", "stosunek"]].head()
```

## Krok 42. Nowe kolumny za pomocą funkcji lambda
Podział na poziom palenia: niski (< 15%), średni (15-30%), wysoki (> 30%).

```python
df["poziom"] = df["val"].apply(lambda x: "niski" if x < 0.15 else ("średni" if x < 0.30 else "wysoki"))
display(df[["location", "year", "val", "poziom"]].head())
df["poziom"].value_counts()
```

## Krok 43. Praca z dużymi plikami – chunksize
Plik czytany jest w kawałkach po 5000 wierszy, bez ładowania całości do pamięci.

```python
suma, liczba = 0, 0
for kawalek in pd.read_csv("wynik_smoking.csv", chunksize=5000):
    suma += kawalek["val"].sum()
    liczba += len(kawalek)

print("Liczba wierszy:", liczba)
print("Średnia val:", suma / liczba)
```
